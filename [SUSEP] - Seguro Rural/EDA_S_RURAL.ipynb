## 3.2 - Join entre as bases

```python
#Variavel com chaves primarias
chaves = ['COD_APO', 'COD_ENDOSSO', 'COD_ITEM']

#Left Join
df_join_lef = df_original_r.merge(
    df_original_s[chaves].drop_duplicates(),
    on=chaves,
    how='left',
    indicator=True
)

df_join_lef['_merge'].value_counts(normalize=True)
```

Notas:

- maioria left_only → apólices sem sinistro 
- parte both → apólices com sinistros
- quase nenhum right_only → consistência
